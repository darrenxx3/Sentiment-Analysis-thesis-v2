```python
from typing import Any
from google_play_scraper import Sort, reviews
from datetime import datetime
import json

class DateTimeEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, datetime):
                return obj.isoformat()
        return super(DateTimeEncoder, self).default(obj)

result1, continuation_token = reviews('com.bca',
                                     lang='en',
                                     country='id',
                                     sort=Sort.NEWEST,
                                     count=20, #banyak rows yang ingin diambil
                                     filter_score_with=None)

result2, continuation_token = reviews('com.bca',
                                     lang='id',
                                     country='id',
                                     sort=Sort.NEWEST,
                                     count=20, #banyak rows yang ingin diambil
                                     filter_score_with=None)

combined_results = result1 + result2

combined_results2, _ = reviews('com.bca',
                    continuation_token=continuation_token)

json_convert2= json.dumps(combined_results2,indent=4, cls=DateTimeEncoder) 
```

```python
print(json_convert2)
```
